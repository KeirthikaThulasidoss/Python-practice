```python
a = 4 # assigning a value to a variable and performing operations 
b = 6

sum = a + b
sub = a - b
mul = a*b
division = a/b
floor_div = a//b
modulo = a%b
exp = a**b

print('Total:', sum)
print('Difference: ',sub)
print('Product: ',mul)
print('Division: ', division)
print('Rounded off: ', floor_div)
print('Remainder: ', modulo)
print('Exponential: ', exp)
```

```python
print('== Addition, subtraction, division, multiplication,modulo division ==')

num_one = 3
num_two = 4

add = num_one+num_two
sub = num_one - num_two
div = num_one/num_two
mul = num_one*num_two
mod = num_one%num_two

print('Total:' ,add)
print('Difference:',sub)
print('Division:' ,div)
print('Multiplication:' ,mul)
print("Remainder: ", mod) #Dividend = (Divisor × Quotient) + Remainder
```
